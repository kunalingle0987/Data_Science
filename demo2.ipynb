```python
import matplotlib.pyplot as plt

years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.plot(years,runs)
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Sachin Runs History")
```

```python
plt.bar(years,runs,label="ViratKohli")
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("virat Runs History Barchart")
plt.legend()
```

```python
import numpy as np
import matplotlib.pyplot as plt

kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 1800]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]
dhoni=[100,400,600,700,900,800,0,1350,1500,1700]

x=np.arange(len(years))
x
```

```python
width=0.25
plt.bar(x-width,kohli,width=width,label="VKohli")
plt.bar(x,sehwag,width=width,label="sehwag")
plt.bar(x+width,dhoni,width=width,label="MsDhoni")

plt.xlabel("Years")
plt.ylabel("Runs")
plt.legend()
plt.tight_layout()
plt.show()
```
