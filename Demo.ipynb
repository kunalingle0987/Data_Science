```python
import matplotlib.pyplot as plt
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
plt.plot(years, runs)
plt.xlabel("Years")
plt.ylabel("Runs Scored")
plt.title("Sachin Tendulkar's Yearly Runs")
plt.show()
```

```python
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.plot(years,kohli,label="Virat kohli")
plt.plot(years,sehwag,label="Viru sehwag")
plt.xlabel("years")
plt.ylabel("Runs")
plt.title("Run_rates")
plt.legend()
```

```python
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.plot(years,kohli,'r*--',label="Virat kohli")
plt.plot(years,sehwag,'mD:',label="Viru sehwag")
plt.xlabel("years")
plt.ylabel("Runs")
plt.title("Run_rates")
plt.legend()
```

```python
import matplotlib.pyplot as plt
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]
dhoni=[100,400,600,700,900,800,0,1350,1500,1900]

plt.plot(years,kohli,'r--*',label="virat Kohli ")
plt.plot(years,sehwag,'h:',label="Viru Sehwag")
plt.plot(years,dhoni,'y-.d',label="ms. Dhoni")

plt.xlabel("Years")
plt.ylabel("RunScored")
plt.title("Cricket_Scored")
plt.legend()
plt.tight_layout()
plt.grid('true')
```
